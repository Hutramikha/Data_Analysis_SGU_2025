# Khám Phá Và Xử Lý Dữ Liệu Marketing Campaign

Notebook này thực hiện khám phá và xử lý dữ liệu từ dataset Customer Personality Analysis trên Kaggle: [https://www.kaggle.com/datasets/imakash3011/customer-personality-analysis](https://www.kaggle.com/datasets/imakash3011/customer-personality-analysis).


## Thư viện cần thiết
- pandas: Xử lý dữ liệu

```python
import pandas as pd
import numpy as np

# Thiết lập hiển thị cho pandas
pd.set_option('display.max_columns', None)
```

## 1. Import Thư Viện Và Nạp Dữ Liệu Vào Notebook

Đọc file CSV và chọn các cột cần thiết.

```python
# Đọc dữ liệu từ file CSV với delimiter là tab
marketing_data = pd.read_csv("./marketing_campaign.csv", delimiter='\t')

# Chọn các cột cần thiết theo yêu cầu
selected_columns = ['ID', 'Year_Birth', 'Education', 'Marital_Status', 'Income', 'Kidhome', 'Teenhome', 
                    'Dt_Customer', 'Recency', 'NumStorePurchases', 'NumWebVisitsMonth']
marketing_data = marketing_data[selected_columns]

# Hiển thị thông tin cơ bản
print('Số hàng và cột:', marketing_data.shape)
print('\nThông tin dữ liệu:')
print(marketing_data.info())
print('\n5 hàng đầu:')
display(marketing_data.head())
```

## 2. Loại Bỏ Dữ Liệu Trùng Lặp

Kiểm tra và loại bỏ duplicates, xóa hàng/cột cụ thể.

```python
# Hiển thị head() để xem dữ liệu gốc
print('Dữ liệu gốc (head()):')
display(marketing_data.head())

# Loại bỏ duplicates trên toàn bộ cột
marketing_data_duplicate = marketing_data.drop_duplicates()
print(f'\nSố hàng sau khi drop duplicates: {len(marketing_data_duplicate)}')

# Xóa một hàng cụ thể tại index 1 (ví dụ)
marketing_data_no_row = marketing_data.drop(labels=[1], axis=0)
print(f'Số hàng sau khi xóa row index 1: {len(marketing_data_no_row)}')

# Xóa một cột cụ thể (ví dụ: Year_Birth)
marketing_data_no_col = marketing_data.drop(labels=['Year_Birth'], axis=1)
print(f'Số cột sau khi xóa cột Year_Birth: {len(marketing_data_no_col.columns)}')
print('\nDữ liệu sau khi xóa cột (head()):')
display(marketing_data_no_col.head())
```

## 3. Thay Thế Dữ Liệu Và Thay Đổi Định Dạng Của Dữ Liệu

Thay thế giá trị trong Teenhome, fill NA trong Income, và thay đổi kiểu dữ liệu.

```python
# Thay thế giá trị trong Teenhome: 0 -> 'has no teen', 1/2 -> 'has teen'
marketing_data['Teenhome_replaced'] = marketing_data['Teenhome'].replace([0, 1, 2], 
                                                                       ['has no teen', 'has teen', 'has teen'])
print('Cột Teenhome_replaced (head()):')
display(marketing_data[['Teenhome', 'Teenhome_replaced']].head())

# Fill NA trong cột Income bằng 0
print(f'\nSố NA trong Income trước: {marketing_data["Income"].isnull().sum()}')
marketing_data['Income'] = marketing_data['Income'].fillna(0)
print(f'Số NA trong Income sau: {marketing_data["Income"].isnull().sum()}')

# Thay đổi kiểu dữ liệu của Income từ float sang int
marketing_data['Income_changed'] = marketing_data['Income'].astype(int)
print('\nKiểu dữ liệu Income_changed:')
print(marketing_data['Income_changed'].dtype)
print('\nIncome_changed (head()):')
display(marketing_data[['Income', 'Income_changed']].head())
```

## 4. Xử Lý Dữ Liệu Thiếu

Kiểm tra và loại bỏ missing values.

```python
# Kiểm tra missing values bằng isnull().sum()
print('Số lượng missing values theo cột:')
missing_summary = marketing_data.isnull().sum()
display(missing_summary)

# Loại bỏ missing values (how='any' để xóa nếu bất kỳ cột nào có NA)
marketing_data_withoutna = marketing_data.dropna(how='any')
print(f'\nShape dữ liệu gốc: {marketing_data.shape}')
print(f'Shape dữ liệu sau dropna: {marketing_data_withoutna.shape}')

# Hiển thị head() của dữ liệu sạch
print('\nDữ liệu sau xử lý missing (head()):')
display(marketing_data_withoutna.head())
```

## 5️ Thống kê mô tả cơ bản

```python
from datetime import datetime

# Lấy năm hiện tại
current_year = datetime.now().year

marketing_data['Age'] = current_year - marketing_data['Year_Birth']

# Kiểm tra thống kê nhanh
print(marketing_data.describe())

# Lấy mean, median, mode của Age
age_mean = marketing_data['Age'].mean()
age_median = marketing_data['Age'].median()
age_mode = marketing_data['Age'].mode()[0]

print("Mean Age:", age_mean)
print("Median Age:", age_median)
print("Mode Age:", age_mode)
```

## Kết Luận

- Dữ liệu đã được tải và chọn cột thành công.
- Đã loại bỏ duplicates và các hàng/cột không cần thiết.
- Thay thế và thay đổi định dạng dữ liệu cho phù hợp.
- Xử lý missing values bằng cách fill và dropna.
