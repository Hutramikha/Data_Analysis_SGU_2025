## 1️⃣ Import thư viện và nạp dữ liệu vào notebook

```python
import pandas as pd
import numpy as np

# Đọc file csv
wine = pd.read_csv("winequality-red.csv")

# Sao chép sang dataframe khác để thao tác
df = wine.copy()

# Xem 10 dòng đầu
df.head(10)

# Kiểm tra kiểu dữ liệu và kích thước tập dữ liệu
print(df.dtypes)
print("Shape:", df.shape)
```

## 2️⃣ Loại bỏ dữ liệu trùng lặp

```python
# Loại bỏ các dòng trùng lặp
df = df.drop_duplicates()

# Kiểm tra kích thước sau khi loại bỏ
print("Shape sau khi loại bỏ duplicates:", df.shape)

# Ví dụ: xóa một dòng cụ thể nếu cần
# df.drop(labels=[1], axis=0)

# Ví dụ: xóa một cột cụ thể nếu cần
# df.drop(labels=['pH'], axis=1)
```

```python
# Kiểm tra lại dữ liệu
df.head()
```

## 3️⃣ Thay thế dữ liệu & thay đổi định dạng dữ liệu

```python
# Kiểm tra các giá trị 0 (nếu có) ở các cột quan trọng
print((df == 0).sum())

# Với một số cột như 'residual sugar' hay 'chlorides',
# nếu có giá trị 0 không hợp lý, có thể thay bằng NaN
cols_replace_zero = ['residual sugar', 'chlorides', 'sulphates', 'alcohol']
df[cols_replace_zero] = df[cols_replace_zero].replace(0, np.nan)

# Tạo thêm cột phân loại chất lượng rượu
df['quality_label'] = pd.cut(df['quality'],
                             bins=[0,5,7,10],
                             labels=['Low','Medium','High'])

# Chuyển kiểu dữ liệu quality về int (nếu cần)
# df['quality'] = df['quality'].astype(int)

# Kiểm tra lại dữ liệu
df.head()
```

## 4️⃣ Xử lý dữ liệu thiếu

```python
# Kiểm tra số lượng giá trị thiếu
print(df.isnull().sum())

# Điền giá trị trung vị cho các cột số bị thiếu
df_filled = df.fillna(df.median(numeric_only=True))

# Hoặc loại bỏ các dòng thiếu
# df_clean = df.dropna(how='any')

print("Shape sau khi xử lý missing:", df_filled.shape)
```

## 5️⃣ Thống kê mô tả cơ bản

```python
# Thống kê nhanh
print(df_filled.describe())

# Lấy mean, median, mode cho cột alcohol
alcohol_mean = df_filled['alcohol'].mean()
alcohol_median = df_filled['alcohol'].median()
alcohol_mode = df_filled['alcohol'].mode()[0]

print("Mean alcohol:", alcohol_mean)
print("Median alcohol:", alcohol_median)
print("Mode alcohol:", alcohol_mode)

# Tính phương sai, độ lệch chuẩn, min, max
alcohol_var = np.var(df_filled['alcohol'])
alcohol_std = np.std(df_filled['alcohol'])
alcohol_min = np.min(df_filled['alcohol'])
alcohol_max = np.max(df_filled['alcohol'])

print("Variance:", alcohol_var)
print("Std dev:", alcohol_std)
print("Min:", alcohol_min)
print("Max:", alcohol_max)

# Percentile & IQR
alcohol_60 = np.percentile(df_filled['alcohol'], 60)
alcohol_iqr = np.percentile(df_filled['alcohol'], 75) - np.percentile(df_filled['alcohol'], 25)

print("60th percentile:", alcohol_60)
print("IQR:", alcohol_iqr)
```

```python
# Lấy mean, median, mode cho cột volatile acidity
va_mean = df_filled['volatile acidity'].mean()
va_median = df_filled['volatile acidity'].median()
va_mode = df_filled['volatile acidity'].mode()[0]

print("Mean volatile acidity:", va_mean)
print("Median volatile acidity:", va_median)
print("Mode volatile acidity:", va_mode)

# Tính phương sai, độ lệch chuẩn, min, max
va_var = np.var(df_filled['volatile acidity'])
va_std = np.std(df_filled['volatile acidity'])
va_min = np.min(df_filled['volatile acidity'])
va_max = np.max(df_filled['volatile acidity'])

print("Variance:", va_var)
print("Std dev:", va_std)
print("Min:", va_min)
print("Max:", va_max)

# Percentile & IQR
va_60 = np.percentile(df_filled['volatile acidity'], 60)
va_iqr = np.percentile(df_filled['volatile acidity'], 75) - np.percentile(df_filled['volatile acidity'], 25)

print("60th percentile:", va_60)
print("IQR:", va_iqr)
```
