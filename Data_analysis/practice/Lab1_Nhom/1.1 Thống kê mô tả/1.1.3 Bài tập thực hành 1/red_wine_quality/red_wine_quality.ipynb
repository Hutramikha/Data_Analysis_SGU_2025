## 1. Định nghĩa vấn đề (Define Problem)

📜 `Wine Dataset` 
| Chỉ số             | Ý nghĩa ngắn gọn |
|------------------------|------------------|
| **Fixed acidity**       | Axit không bay hơi (malic, lactic, tartaric…), đặc trưng cơ bản của rượu. |
| **Volatile acidity**    | Axit bay hơi (chủ yếu acetic). Quá cao gây mùi giấm. |
| **Citric acid**         | Axit hữu cơ nhẹ, tạo độ “tươi” cho rượu. |
| **Residual sugar**      | Đường còn lại sau lên men (g/L), quyết định độ ngọt. |
| **Chlorides**           | Hàm lượng muối (ion Cl⁻) trong rượu. |
| **Free sulfur dioxide** | SO₂ tự do, có tác dụng kháng khuẩn & chống oxi hóa. |
| **Total sulfur dioxide**| Tổng SO₂ (tự do + liên kết). |
| **Density**             | Mật độ rượu so với nước, phụ thuộc cồn & đường. |
| **pH**                  | Độ axit/kiềm (0–14), rượu thường 3–4. |
| **Sulphates**           | Phụ gia góp phần vào mức SO₂, bảo quản rượu. |
| **Alcohol**             | % cồn của rượu. |
| **Quality**             | Điểm đánh giá chất lượng cảm quan (3–8). |

### 4.4 Boxplot

```python
columns = [
    "fixed_acidity",
    "volatile_acidity",
    "citric_acid",
    "residual_sugar",
    "chlorides",
    "free_sulfur_dioxide",
    "total_sulfur_dioxide",
    "density",
    "pH",
    "sulphates",
    "alcohol"
]

fig, axes = plt.subplots(3, 4, figsize=(16, 10))
axes = axes.flatten()

for i, col in enumerate(columns):
    sns.boxplot(y=df[col], ax=axes[i])
    axes[i].set_title(f'Boxplot của {col}')

# Ẩn subplot dư thừa nếu có
for j in range(len(columns), len(axes)):
    axes[j].set_visible(False)

plt.tight_layout()
plt.show()

# Tính Median, IQR và số lượng outlier cho từng biến
print("\nThống kê Median, IQR và số lượng Outliers:")
for col in columns:
    median = df[col].median()
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    outliers = ((df[col] < q1 - 1.5 * iqr) | (df[col] > q3 + 1.5 * iqr)).sum()
    print(f'{col}: Median = {median:.2f}, IQR = {iqr:.2f}, Outliers = {outliers}')
```

**Chi tiết từng biến**  

- **fixed_acidity**: Median = 7.9; IQR = 2.1; Outliers = 41 → phân bố tập trung, có một số giá trị acid cố định cao/thấp bất thường.  
- **volatile_acidity**: Median = 0.52; IQR = 0.25; Outliers = 19 → khá gọn, ít ngoại lai.  
- **citric_acid**: Median = 0.26; IQR = 0.34; Outliers = 1 → ổn định, gần như không có ngoại lai.  
- **residual_sugar**: Median = 2.2; IQR = 0.7; Outliers = 126 → đuôi dài, nhiều mẫu đường dư khác thường.  
- **chlorides**: Median = 0.08; IQR = 0.02; Outliers = 87 → nhiều mẫu muối cao bất thường.  
- **free_sulfur_dioxide**: Median = 14.0; IQR = 14.0; Outliers = 26 → phân tán mạnh, có điểm SO₂ tự do cao bất thường.  
- **total_sulfur_dioxide**: Median = 38.0; IQR = 41.0; Outliers = 45 → SO₂ tổng phân tán mạnh.  
- **density**: Median = 1.00; IQR ≈ 0.00; Outliers = 35 → mật độ hầu như sát 1, nhưng có vài điểm khác biệt.

**Kết luận **

- Một số biến có **nhiều ngoại lai** như `residual_sugar`, `chlorides`, `total_sulfur_dioxide`.
