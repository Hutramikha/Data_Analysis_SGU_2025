## 2. Sử Dụng Công Cụ Ydata Profiling

Đọc dữ liệu và tạo báo cáo EDA tự động, lưu vào file HTML.

```python
import pandas as pd
from ydata_profiling import ProfileReport

# Đọc dữ liệu từ file CSV
marketing_data = pd.read_csv("./marketing_campaign.csv")

# Hiển thị thông tin cơ bản về dữ liệu
print('Số hàng và cột:', marketing_data.shape)
print('\n5 hàng đầu:')
display(marketing_data.head())
```

```python
# Tạo báo cáo EDA tự động sử dụng ProfileReport
profile = ProfileReport(marketing_data, title="Customer Personality Analysis EDA Report")

# Lưu báo cáo vào file HTML
profile.to_file("./profile_output.html")

print('Báo cáo đã được lưu vào: ./profile_output.html')
```

### Nhận Xét
- Báo cáo bao quát toàn bộ dataset (29 cột, 2240 hàng), phát hiện missing values (Income: 24 NA), outliers, và tương quan (e.g., Income vs. MntWines).
- Thời gian tạo báo cáo khoảng 10-20 giây tùy kích thước dữ liệu.

### Kết Luận
Ydata_profiling tiết kiệm thời gian so với EDA thủ công, cung cấp insights nhanh chóng về phân bố, tương quan.

## 3. Tiến Hành EDA Trên Trang Tập Tin Profile_Output.html

Mở file `./profile_output.html` trong trình duyệt để xem báo cáo tương tác. Dưới đây là tóm tắt các insights chính từ báo cáo.

#### Các Insights Chính Từ Báo Cáo:
- **Overview**: 2240 quan sát, 29 biến (20 numerical, 9 categorical). 1.07% missing values chủ yếu ở Income.
- **Variables**: 
  - Income: Phân bố lệch phải (mean ~52k, median ~51k), outliers cao (>100k).
  - Education: Phân loại (PhD ~9%, Master ~22%).
  - MntWines: Tương quan cao với Income (0.68).
- **Correlations**: Ma trận tương quan hiển thị mối liên hệ giữa chi tiêu (Mnt*) và Income.
- **Missing Values**: Income có 24 NA, có thể impute bằng median.
- **Interactions**: Biểu đồ scatter cho thấy clusters khách hàng theo chi tiêu và thu nhập.

### Nhận Xét
- Báo cáo tương tác cho phép zoom, filter, giúp khám phá sâu (e.g., histogram cho từng biến).
- Phát hiện vấn đề: High cardinality ở Dt_Customer (ngày khách hàng), cần xử lý.
- Ưu điểm: Tự động, toàn diện; nhược điểm: Không tùy chỉnh sâu.
