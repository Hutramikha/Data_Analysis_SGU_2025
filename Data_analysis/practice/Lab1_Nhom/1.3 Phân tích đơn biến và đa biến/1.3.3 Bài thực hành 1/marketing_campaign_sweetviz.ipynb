# Phân tích Dữ liệu Khám phá (EDA) với SweetViz trên Tập Dữ liệu Marketing Campaign

## Giới thiệu về SweetViz

SweetViz là một thư viện Python mã nguồn mở được thiết kế để tự động hóa quá trình phân tích dữ liệu khám phá (Exploratory Data Analysis - EDA). Nó giúp tạo ra các báo cáo trực quan hóa chi tiết dưới dạng file HTML, bao gồm:

### Các Tính Năng Chính:
- **Phân tích đơn biến (Univariate Analysis)**: Phân phối, giá trị thiếu, kiểu dữ liệu.
- **Phân tích nhị biến (Bivariate Analysis)**: Tương quan giữa các biến, so sánh phân phối.
- **Ma trận tương quan (Correlation Matrix)**: Hiển thị mối quan hệ giữa các biến số.
- **Phân loại biến**: Tự động nhận diện biến số và biến phân loại.
- **So sánh dữ liệu**: Có thể so sánh hai tập dữ liệu (ví dụ: train vs test).
- **Trực quan hóa**: Biểu đồ histogram, boxplot, heatmap, word clouds (cho text), v.v.
- **Hiệu suất cao**: Xử lý nhanh dữ liệu lớn mà không cần cấu hình phức tạp.

SweetViz đặc biệt hữu ích cho việc khám phá nhanh dữ liệu mà không cần viết nhiều mã thủ công.

Hoặc trong notebook (sử dụng cell code bên dưới).

## Tập Dữ liệu Marketing Campaign

Tập dữ liệu này chứa thông tin về khách hàng của một công ty bán lẻ, bao gồm:
- **ID**: Mã khách hàng.
- **Year_Birth**: Năm sinh.
- **Education**: Trình độ học vấn.
- **Marital_Status**: Tình trạng hôn nhân.
- **Income**: Thu nhập hàng năm.
- **Kidhome/Teenhome**: Số lượng trẻ em nhỏ/lớn.
- **Dt_Customer**: Ngày đăng ký.
- **Recency**: Số ngày kể từ lần mua cuối.
- **MntWines/MntFruits/...**: Số tiền chi tiêu cho các hạng mục sản phẩm.
- **NumDealsPurchases/...**: Số lượng giao dịch.
- **AcceptedCmp1-5**: Chấp nhận chiến dịch marketing.
- **Complain**: Khiếu nại.
- **Response**: Phản hồi chiến dịch cuối.



```python
import pandas as pd
import sweetviz as sv
from io import StringIO

# Dữ liệu CSV được cung cấp (phần truncated, bạn có thể thay bằng file đầy đủ)
csv_data = '''ID	Year_Birth	Education	Marital_Status	Income	Kidhome	Teenhome	Dt_Customer	Recency	MntWines	MntFruits	MntMeatProducts	MntFishProducts	MntSweetProducts	MntGoldProds	NumDealsPurchases	NumWebPurchases	NumCatalogPurchases	NumStorePurchases	NumWebVisitsMonth	AcceptedCmp3	AcceptedCmp4	AcceptedCmp5	AcceptedCmp1	AcceptedCmp2	Complain	Z_CostContact	Z_Revenue	Response
5524	1957	Graduation	Single	58138	0	0	04-09-2012	58	635	88	546	172	88	88	3	8	10	4	7	0	0	0	0	0	0	3	11	1
2174	1954	Graduation	Single	46344	1	1	08-03-2014	38	11	1	6	2	1	6	2	1	1	2	5	0	0	0	0	0	0	3	11	0
# ... (thêm toàn bộ dữ liệu CSV ở đây để hoàn chỉnh, vì bị truncated trong prompt)
'''

# Đọc dữ liệu từ string (hoặc thay bằng pd.read_csv('marketing_campaign.csv') nếu có file)
df = pd.read_csv(StringIO(csv_data), sep='\t')

# Hiển thị thông tin cơ bản về dữ liệu
print(df.head())
print(f'Shape: {df.shape}')
print(df.info())
```

## Tạo Báo Cáo EDA với SweetViz

Sử dụng `sweetviz.analyze()` để tạo báo cáo tự động. Báo cáo sẽ được lưu dưới dạng HTML và mở trong trình duyệt.

```python
# Tạo báo cáo EDA
report = sv.analyze(df)

# Hiển thị báo cáo trong HTML (mở file report.html trong trình duyệt)
report.show_html('marketing_campaign_report.html')

print('Báo cáo đã được tạo: marketing_campaign_report.html')
```

## Giải Thích Báo Cáo

Khi mở file `marketing_campaign_report.html`:
- **Overview**: Tổng quan về dữ liệu (số lượng hàng, cột, giá trị thiếu).
- **Variables**: Chi tiết từng biến (phân phối, outliers).
- **Associations**: Tương quan giữa các biến (heatmap, biểu đồ đôi).
- **Missing Values**: Biểu đồ giá trị thiếu.

### Cách Sử Dụng Nâng Cao
- **So sánh hai tập dữ liệu**: `sv.compare(df_train, df_test)`.
- **Tùy chỉnh**: `sv.analyze(df, target_feat='Response')` để tập trung vào biến mục tiêu.
- **Lưu báo cáo**: `report.show_html(open_browser=False)` để chỉ lưu file mà không mở.

SweetViz giúp tiết kiệm thời gian EDA, đặc biệt với tập dữ liệu marketing này để xác định các mẫu chi tiêu, phản hồi campaign, v.v.

```python
# Ví dụ nâng cao: Phân tích với biến mục tiêu 'Response'
df_no_nan_target = df[df['Response'].notna()]
report_target = sv.analyze(df_no_nan_target, target_feat='Response')
report_target.show_html('marketing_campaign_target_report.html')
print('Báo cáo với target: marketing_campaign_target_report.html')
```
